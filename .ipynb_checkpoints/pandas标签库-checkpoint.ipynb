```python
import pandas as pd
import numpy as np
```

```python
#创建字典对象
dit_v={'a':0,'b':1,'c':3}
dit_v
```

用列表创建对象

```python
v=[1,2,3,4,5]
k=['a','b','c','d','e']
sr=pd.Series(v,index=k)#v值 k键
sr
```

```python
v=[33,43,26,52]
i=['1号','2号','3号','4号']
sr1=pd.Series(v,index=i)
sr1
```

```python
v1=['男','女','男','女']
i1=['1号','2号','3号','4号']
sr2=pd.Series(v1,index=i1)
sr2
```

```python
#创建df对象
df=pd.DataFrame({'年龄':sr1,'性别':sr2})
df
```

```python
a=['1','2']
b=['zhus','kk']
arr=pd.Series(b,index=a)
arr
```

```python
a1=['1','2']
b2=['24','27']
arr2=pd.Series(b2,index=a1)
df2=pd.DataFrame({'年龄':arr2,'姓名':arr})
df2
```
