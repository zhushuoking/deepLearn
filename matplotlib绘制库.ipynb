```python
fig=plt.figure()#创建图窗
x=[1,2,3,4,5]
y=[1,8,27,64,66]
x1=[2,3,6,7]
y1=[2,4,6,0]
plt.subplot(3,1,1),plt.plot(x,y)#(3,1,1)表示三行一列的第一个位置
plt.subplot(3,1,2),plt.plot(x,y)#(3,1,2)表示三行一列的第二个位置
plt.subplot(3,1,3),plt.plot(x1,y1)#(3,1,3)表示三行一列的第三个位置
```
