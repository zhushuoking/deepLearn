二维数组,三维数组

```python
#创建一维数组向量
arr3=np.array([1,2,3])
print(arr3)
#创建二维数组行矩阵
arr4=np.array([[1,2,3]])
print(arr4)
#创建二维数组列矩阵
arr5=np.array([[1],[2],[3]])
print(arr5)
```

np.arange(x,y,z)  x:从零开始到x之前的数  y:从x开始到y之前结束  z:步长

```python
arr6=np.arange(10,20,3)#从10开始到20之前结束步长为3
print(arr6)
```
