```python
import numpy as np
#创造随机数组
```

```python
#0-1均匀分布的浮点型随机数组
arr1=np.random.random((2,5))*100#二行五列数组
arr2=np.random.random((6))
print(arr1)
print(arr2)
```

```python
#整数型随机数
arr3=np.random.randint(1,100,(2,5))
print(arr3)
```

```python
#服从正态分布数组
arr4=np.random.normal(0,1,(2,5))#均值为0，标准差为1，二行五列数组
print(arr4)
```
